```python
import matplotlib.pyplot as plt

# Create figure
fig, ax = plt.subplots(figsize=(8, 10))
ax.axis('off')

# Define boxes and positions (x, y)
boxes = {
    "Start": (0.5, 0.95, "Start / Home Page"),
    "Login": (0.5, 0.87, "User Registration / Login"),
    "Apply": (0.5, 0.79, "User Applies for E-Pass"),
    "ToAdmin": (0.5, 0.71, "Application Sent to Admin"),
    "Approve": (0.25, 0.63, "Admin Approves Pass"),
    "Reject": (0.75, 0.63, "Admin Rejects Pass"),
    "Generate": (0.25, 0.55, "System Generates E-Pass\n(With QR Code)"),
    "Notice": (0.75, 0.55, "User Gets Rejection Notice"),
    "Download": (0.25, 0.47, "User Downloads / Views Pass"),
    "Verify": (0.25, 0.39, "Verification at Checkpoint\n(QR Code Scanned)"),
    "Access": (0.25, 0.31, "Access Granted / Denied")
}

# Draw boxes
for key, (x, y, text) in boxes.items():
    ax.text(x, y, text, ha='center', va='center', fontsize=9, fontweight='bold',
            bbox=dict(facecolor='lightyellow', edgecolor='black', boxstyle='round,pad=0.5'))

# Draw connecting arrows
def arrow(start, end):
    x1, y1 = boxes[start][0], boxes[start][1] - 0.03
    x2, y2 = boxes[end][0], boxes[end][1] + 0.03
    ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle="->", color="black", lw=1.5))

# Main flow arrows
arrow("Start", "Login")
arrow("Login", "Apply")
arrow("Apply", "ToAdmin")
arrow("ToAdmin", "Approve")
arrow("ToAdmin", "Reject")
arrow("Approve", "Generate")
arrow("Generate", "Download")
arrow("Download", "Verify")
arrow("Verify", "Access")
arrow("Reject", "Notice")

plt.title("Flowchart – Online E-Pass Management System", fontsize=14, fontweight='bold', pad=20)
plt.tight_layout()
plt.show()
```
